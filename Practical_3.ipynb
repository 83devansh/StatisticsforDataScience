**Q1. A. Identify all duplicate cases using prof.**

```python
duplicates = df[df.duplicated(subset='prof', keep=False)]

print(duplicates['prof'].unique())
```

```python
print(duplicates)
```

**B. Average and standard deviation of age using all observations**

```python
print("Mean age:", df['age'].mean())
print("Standard deviation:", df['age'].std())
```

**C. Keep only one observation for each instructor**

```python
df_unique = df.drop_duplicates(subset='prof', keep='first')

print("Number of observations:", len(df_unique))
```

```python
print("Mean age:", df_unique['age'].mean())
print("Standard deviation:", df_unique['age'].std())
```

**Q2. Bar chart: Lower-division courses vs teaching evaluations**

```python
print(df.columns)
```

```python
import matplotlib.pyplot as plt
```

```python
avg_eval = df.groupby('division')['eval'].mean()

print(avg_eval)

avg_eval.plot(kind='bar')

plt.xlabel("Course Division")
plt.ylabel("Average Teaching Evaluation")
plt.title("Average Teaching Evaluation by Course Division")
plt.xticks(rotation=0)
plt.show()
```

**Conclusion:**

**Yes. Instructors teaching lower-division courses receive higher average teaching evaluations than instructors teaching upper-division courses.**

**Q3. Relationship Between Age and Teaching Evaluation**

```python
import seaborn as sns
```

```python
sns.scatterplot(data=df, x='age', y='eval')

plt.xlabel("Age")
plt.ylabel("Teaching Evaluation")
plt.title("Age vs Teaching Evaluation")
plt.show()
```

```python
print("Correlation:", df['age'].corr(df['eval']))
```

**The correlation between age and teaching evaluation is -0.0517. This indicates a very weak negative relationship between age and teaching evaluation. Since the correlation is very close to zero, there is almost no linear relationship between the two variables.**

**Q4. Age vs Teaching Evaluation by Gender**

```python
sns.scatterplot(
    data=df,
    x='age',
    y='eval',
    hue='gender'
)

plt.xlabel("Age")
plt.ylabel("Teaching Evaluation")
plt.title("Age vs Teaching Evaluation by Gender")
plt.show()
```

**The scatter plot shows the relationship between age and teaching evaluation scores for male and female instructors. The points for both genders are widely scattered, indicating that there is no strong relationship between age and teaching evaluation. Both male and female instructors have a wide range of evaluation scores across different ages.**

**Q5. Age vs Evaluation by Gender AND Tenure**

```python
sns.scatterplot(
    data=df,
    x='age',
    y='eval',
    hue='gender',
    style='tenure'
)

plt.xlabel("Age")
plt.ylabel("Teaching Evaluation")
plt.title("Age vs Teaching Evaluation by Gender and Tenure")
plt.show()
```

**The scatter plot shows the relationship between age and teaching evaluation scores while differentiating instructors by gender and tenure. The evaluation scores are widely distributed across different ages for both genders and tenure groups. Therefore, there does not appear to be a strong relationship between age and teaching evaluation based on the scatter plot.**

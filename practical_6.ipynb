**Q1. T-Test**
**Question:**

**Using the teachers' rating dataset, does gender affect teaching evaluation rates?**

```python
import pandas as pd
from scipy.stats import ttest_ind
```

```python
df = pd.read_csv("teachingratings (1).csv")
```

```python
male = df[df["gender"] == "male"]["eval"]
female = df[df["gender"] == "female"]["eval"]
```

```python
t_stat, p_value = ttest_ind(male, female, equal_var=False)

print("T-statistic:", t_stat)
print("P-value:", p_value)

# Decision
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

**Conclusion**

**Since the p-value is less than 0.05, we reject the null hypothesis. Therefore, there is a statistically significant difference in teaching evaluation rates between male and female instructors. This indicates that gender affects teaching evaluation rates.**

**Q2. ANOVA
Question:**

**Using the teachers' rating dataset, does beauty score for instructors differ by age?**

```python
df["age_group"] = pd.cut(
    df["age"],
    bins=[0, 30, 40, 50, 100],
    labels=["Under 30", "30-40", "40-50", "50+"]
)
```

```python
groups = [
    group["beauty"].dropna()
    for name, group in df.groupby("age_group", observed=True)
]
```

```python
from scipy.stats import f_oneway
```

```python
f_stat, p_value = f_oneway(*groups)

print("F-statistic:", f_stat)
print("P-value:", p_value)

# Decision
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

**Conclusion**

**Since the p-value is less than 0.05, we reject the null hypothesis. Therefore, there is a statistically significant difference in the beauty scores of instructors across the different age groups. This indicates that beauty score differs by age.**

**Q3. Chi-Square Test
Question:**

**Using the teachers' rating dataset, is there an association between tenure and gender?**

```python
from scipy.stats import chi2_contingency
```

```python
table = pd.crosstab(df["tenure"], df["gender"])

print("Contingency Table:")
print(table)
```

```python
chi2, p_value, dof, expected = chi2_contingency(table)

print("\nChi-square statistic:", chi2)
print("Degrees of freedom:", dof)
print("P-value:", p_value)

# Decision
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

**Conclusion:**

**Since the p-value is less than 0.05, we reject the null hypothesis. Therefore, there is a statistically significant association between tenure and gender. This indicates that tenure and gender are associated.**

**Q4. Correlation
Question:**

**Using the teachers' rating dataset, is teaching evaluation score correlated with beauty score?**

```python
from scipy.stats import pearsonr
```

```python
data = df[["eval", "beauty"]].dropna()
```

```python
correlation, p_value = pearsonr(data["eval"], data["beauty"])

print("Correlation coefficient:", correlation)
print("P-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

**Conclusion**

Conclusion:

Since the p-value is less than 0.05, we reject the null hypothesis. Therefore, there is a statistically significant correlation between teaching evaluation score and beauty score. This indicates that teaching evaluation score is correlated with beauty score.

**Q1. Probability of receiving an evaluation score greater than 4.5**

```python
import pandas as pd
```

```python
df = pd.read_csv("teachingratings (1).csv")
```

```python
total = len(df)
```

```python
df.head()
```

```python
df.columns
```

```python
count = (df["eval"] > 4.5).sum()
```

```python
probability = count / total
```

```python
print("Total observations:", total)
print("Number of scores greater than 4.5:", count)
print("Probability:", probability)
```

**Conclusion:**

**Therefore, the probability of receiving an evaluation score greater than 4.5 is [your Q1 probability]. This means there is approximately [probability × 100]% chance of receiving an evaluation score greater than 4.5.**

**QUE.2 Probability of evaluation score greater than 3.5 and less than 4.2**

**Q3.Using the two-tailed test from a normal distribution**

```python
import math
from scipy.stats import norm
```

```python
mu = 12
sigma = 5.5
n = 36
sample_mean = 10.7
alpha = 0.05
```

```python
standard_error = sigma / math.sqrt(n)
```

```python
z_score = (sample_mean - mu) / standard_error
```

```python
p_value = 2 * (1 - norm.cdf(abs(z_score)))

print("Standard Error:", standard_error)
print("Z-score:", z_score)
print("P-value:", p_value)
```

```python
if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

**Conclusion**

**Since p-value (0.1560) > α (0.05), we fail to reject the null hypothesis.**

**Therefore, there is insufficient evidence to conclude that the mean points scored by the regional players is different from the historic mean of 12 points per game.**
